```python
# dernier essai de S1 pour chaque utilisateur
last_s1 = (
    sess1.sort_values(["user_id", "order_id"])
         .groupby("user_id")
         .tail(1)
         [["user_id", "timestamp", "RT"]]
         .copy()
)
last_s1["end_s1"] = (
    last_s1["timestamp"]
    + pd.to_timedelta(last_s1["RT"], unit="s")
)

# début de S2
last_s1["start_s2"] = (
    last_s1["end_s1"] + pd.Timedelta(days=7)
)
```

```python
sess2 = sess2.merge(
    last_s1[["user_id", "start_s2"]],
    on="user_id",
    how="left"
)
```

```python
# 1. Trier S2 dans l'ordre des essais
sess2 = sess2.sort_values(
    ["user_id", "order_id"]
).copy()

# 2. RT numérique
sess2["RT"] = pd.to_numeric(
    sess2["RT"],
    errors="coerce"
)

# 3. Remplacer les RT manquants
sess2["RT"] = sess2["RT"].fillna(sess2["RT"].median())

# 4. Temps relatif au début de S2
sess2["time_seconds"] = (
    sess2.groupby("user_id")["RT"]
         .transform(
             lambda x: x.cumsum().shift(fill_value=0)
         )
)

# 5. Timestamp réel/reconstruit de chaque essai S2
sess2["timestamp"] = (
    sess2["start_s2"]
    + pd.to_timedelta(sess2["time_seconds"], unit="s")
)
```

```python
sess2.columns
```

```python
sess2=sess2[["user_id","Stim_ID","skill_name","correct","Session_Number","Learning_Condition_Short_Label",'Trial_N',"order_id","RT","timestamp","Assigned_Criterion"]]
```

```python
sess2=sess2.rename(columns={"Stim_ID":"item_id","skill_name":"KC"})
```
